```python
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score
```

```python
iris = load_iris()
```

```python
print(iris.feature_names)
```

```python
print(iris.target_names)
```

```python
df = pd.DataFrame(
    iris.data,
    columns=iris.feature_names
)

df["species"] = iris.target_names[iris.target]
df.head()
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
df.info()
```

```python
df.describe()
```

```python
print(df["species"].unique())
```

```python
print(df["species"].value_counts())
```

```python
plt.figure(figsize=(8,6))

plt.scatter(
    df["sepal length (cm)"],
    df["sepal width (cm)"]
)

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Sepal Length vs Sepal Width")

plt.show()
```

```python
X = iris.data
y = iris.target
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training observations:", len(X_train))
print("Testing observations:", len(X_test))
```

```python
model = KNeighborsClassifier(n_neighbors=3)
```

```python
plt.figure(figsize=(8,6))

for species in df["species"].unique():
    subset = df[df["species"] == species]

    plt.scatter(
        subset["sepal length (cm)"],
        subset["sepal width (cm)"],
        label=species
    )

plt.xlabel("Sepal Length (cm)")
plt.ylabel("Sepal Width (cm)")
plt.title("Iris Species Based on Sepal Measurements")
plt.legend()

plt.show
```

```python
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred)
```

```python
print("Actual: ")
print(y_test)

print("\nPredicted: ")
print(y_pred)
```

```python
accuracy = accuracy_score(y_test, y_pred)
print("Accuracy: ", accuracy)
```

```python
new_flower = [[5.1, 3.5, 1.4, 0.2]]
```

```python
prediction = model.predict(new_flower)
```

```python
print(
    "Predicted species: ",
    iris.target_names[prediction[0]]
)
```

```python
probabilities = model.predict_proba(new_flower)
print(probabilities)
```

```python
for species, probability in zip(
    iris.target_names,
    probabilities[0]
):
    print(f"{species}: {probability:.2%}")
```

```python
k_values = [1,3,5,7,9]
accuracies = []

for k in k_values:
    model = KNeighborsClassifier(n_neighbors=k)
    model.fit(X_train, y_train)
    y_pred = model.predict(X_test)
    accuracy = accuracy_score(y_test, y_pred)
    accuracies.append(accuracy)
    print(f"K={k}, Accuracy={accuracy:.2f}")
```

```python
plt.figure(figsize=(8,5))

plt.plot(
    k_values,
    accuracies,
    marker="o"
)

plt.xlabel("K Value")
plt.ylabel("Accuracy")
plt.title("KNN Model Accuracy for Different K values")

plt.xticks(k_values)
plt.show()
```
